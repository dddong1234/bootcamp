```python
import os

from IPython.core.magic import output_can_be_silenced
from dotenv import load_dotenv
import json, re
from openai import OpenAI
import pandas as pd
import numpy as np
import gradio as gr

load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")
client = OpenAI(api_key=api_key)

def profile(name, age, subscribe):
    msg = f"{name}님, {age}세"
    note = "구독 중" if subscribe else "미구독"
    return msg, note

demo = gr.Interface(
    fn=profile,
    inputs = [
        gr.Textbox(label="이름"),
        gr.Slider(0,100, label = "나이"),
        gr.Checkbox(label="뉴스레터 구독")
    ],
    outputs = [
        gr.Textbox(label="소개"),
        gr.Textbox(label="구독 상태")
    ]
)

demo.launch()
```
